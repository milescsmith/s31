```python
class NetworkInterfaceAssociation(NamedTuple):
    IpOwnerId: str
    PublicDnsName: ParseResult
    PublicIp: IPv4Address

AttachmentStatus = Literal["attached", "detached"]

class NetworkInterfaceAttachment(NamedTuple):
    AttachTime: datetime
    AttachmentId: str
    DeleteOnTermination: bool
    DeviceIndex: int
    Status: AttachmentStatus
    NetworkCardIndex: int

class NetworkInterfaceAttachmentGroup(NamedTuple):
    GroupName: str
    GroupId: str

class PrivateIPAddress(NamedTuple):
    Association: NetworkInterfaceAssociation
    Primary: bool
    PrivateDnsName: str
    PrivateIpAddress: IPv4Address

class MACAddress():
    def __init__(self, address: str):
        if self.validate(address):
            self.address = address

    def validate(self, address) -> bool:
        return True if re.match("[0-9a-f]{2}([-:]?)[0-9a-f]{2}(\\1[0-9a-f]{2}){4}$", address.lower()) else False

    def __repr__(self) -> str:
        return self.address

@dataclass
class NetworkInterface:
    Association: NetworkInterfaceAssociation = None
    Attachment: NetworkInterfaceAttachment = None
    Description: str = None
    Groups: list[NetworkInterfaceAttachmentGroup] = None
    Ipv6Addresses: IPv6Address = None
    MacAddress: MACAddress = None
    NetworkInterfaceId: str = None
    OwnerId: str = None
    PrivateDnsName: str = None
    PrivateIpAddress: IPv4Address = None
    PrivateIpAddresses: list[PrivateIPAddress] = None
    SourceDestCheck: bool = None
    Status: str = None
    SubnetId: str = None
    VpcId: str = None
    InterfaceType: str = None

    @classmethod
    def from_response(cls, response):
        ni = cls()
        
        ni.Association = NetworkInterfaceAssociation(
            response["Association"]["IpOwnerId"],
            urlparse(response["Association"]["PublicDnsName"]), 
            response["Association"]["PublicIp"]
        )
        ni.Attachment = NetworkInterfaceAttachment(
            AttachTime=response["Attachment"]["AttachTime"],
            AttachmentId=response["Attachment"]["AttachmentId"],
            DeleteOnTermination=response["Attachment"]["DeleteOnTermination"],
            DeviceIndex=response["Attachment"]["DeviceIndex"],
            Status=response["Attachment"]["Status"],
            NetworkCardIndex=response["Attachment"]["NetworkCardIndex"],
        )
        ni.Description = response["Description"]
        ni.Groups = [
            NetworkInterfaceAttachmentGroup(GroupName, GroupId) 
            for GroupName, GroupId 
            in response["Groups"]
        ],
        ni.Ipv6Addresses = [
            IPv6Address(_) 
            for _
            in response["Ipv6Addresses"]
        ],
        ni.MacAddress = MACAddress(response["MacAddress"])
        ni.NetworkInterfaceId = response["NetworkInterfaceId"]
        ni.OwnerId = response["OwnerId"]
        ni.PrivateDnsName = response["PrivateDnsName"]
        ni.PrivateIpAddress = IPv4Address(response["PrivateIpAddress"])
        ni.PrivateIpAddresses = [
            PrivateIPAddress(
                Association = NetworkInterfaceAssociation(
                    IpOwnerId = _["Association"]["IpOwnerId"],
                    PublicDnsName = _["Association"]["PublicDnsName"],
                    PublicIp = _["Association"]["PublicIp"],
                ),
                Primary = _["Primary"],
                PrivateDnsName = _["PrivateDnsName"],
                PrivateIpAddress = IPv4Address(_["PrivateIpAddress"]),)
            for _
            in response["PrivateIpAddresses"]
        ]
        ni.SourceDestCheck = response["SourceDestCheck"]
        ni.Status = response["Status"]
        ni.SubnetId = response["SubnetId"]
        ni.VpcId = response["VpcId"]
        ni.InterfaceType = response["InterfaceType"]

        return ni
```

```python
class InstanceMonitor(NamedTuple):
    State: Literal["enabled", "disabled"]

class InstancePlacement(NamedTuple):
    AvailabilityZone: str
    GroupName: str
    Tenancy: str

class InstanceState(NamedTuple):
    Code: int
    Name: str

class InstanceBDMEbs(NamedTuple):
    AttachTime: datetime
    DeleteOnTermination: bool
    Status: Literal["attached","detached"]
    VolumeId: str
    
class InstanceBlockDeviceMapping(NamedTuple):
    DeviceName: str
    Ebs: InstanceBDMEbs

class InstanceIamInstanceProfile(NamedTuple):
    Arn: str
    Id: str

class InstanceSecurityGroup(NamedTuple):
    GroupName: str
    GroupId: str

class Instance:
    def __init__(self, response: dict[str, Any]):
        self.AmiLaunchIndex: int = response["AmiLaunchIndex"]
        self.ImageId: str = response["ImageId"]
        self.InstanceId: str = response["InstanceId"]
        self.InstanceType: str  = response["InstanceType"]# really, should probably be a Literal, but I'm not making the list of options
        self.KeyName: str = response["KeyName"]
        self.LaunchTime: datetime = response["LaunchTime"]
        self.Monitoring: dict[str, str] = InstanceMonitor(**response["Monitoring"])
        self.Placement: dict[str, str] = InstancePlacement(**response["Placement"])
        self.Platform: str = response["Platform"]
        self.PrivateDnsName: str = urlparse(response["PrivateDnsName"])
        self.PrivateIpAddress: str  = IPv4Address(response["PrivateIpAddress"])
        self.ProductCodes: list[Any] = response["ProductCodes"]
        self.PublicDnsName: str = urlparse(response["PublicDnsName"])
        self.PublicIpAddress: str = IPv4Address(response["PublicIpAddress"])
        self.State: dict[str, str | int] = InstanceState(**response["State"])
        self.StateTransitionReason: Any = response["StateTransitionReason"]
        self.SubnetId: str = response["SubnetId"]
        self.VpcId: str = response["VpcId"]
        self.Architecture: str = response["Architecture"] # another Literal or Enum
        self.BlockDeviceMappings: list[dict[str, Any]] = [
            InstanceBlockDeviceMapping(
                DeviceName=_["DeviceName"],
                Ebs=InstanceBDMEbs(**_["Ebs"])
            )
            for _
            in response["BlockDeviceMappings"]
        ]
        self.ClientToken: str = response["ClientToken"]
        self.EbsOptimized: bool = response["EbsOptimized"]
        self.EnaSupport: bool = response["EnaSupport"]
        self.Hypervisor: str = response["Hypervisor"]
        self.IamInstanceProfile: dict[str, str] = InstanceIamInstanceProfile(**response["IamInstanceProfile"])
        self.NetworkInterfaces: list[NetworkInterface] = [
            NetworkInterface(_)
            for _ in
            response["NetworkInterfaces"]
        ]
        self.RootDeviceName: str = response["RootDeviceName"]
        self.RootDeviceType: str = response["RootDeviceType"]
        self.SecurityGroups: list[dict[str, str]] = [
            InstanceSecurityGroup(**_)
            for _
            in response["SecurityGroups"]
        ]
        self.SourceDestCheck: bool = response["SourceDestCheck"]
        self.Tags: list[dict[str, str]] = response["Tags"]
        self.VirtualizationType: str = response["VirtualizationType"]
        self.CpuOptions: dict[str, int] = response["CpuOptions"]
        self.CapacityReservationSpecification: dict[str, str] = response["CapacityReservationSpecification"]
        self.HibernationOptions: dict[str, bool] = response["HibernationOptions"]
        self.MetadataOptions: dict[str, str] = response["MetadataOptions"]
        self.EnclaveOptions: dict[str, bool] = response["EnclaveOptions"]
        self.PlatformDetails: str = response["PlatformDetails"]
        self.UsageOperation: str  = response["UsageOperation"]#?
        self.UsageOperationUpdateTime: datetime = response["UsageOperationUpdateTime"]
        self.PrivateDnsNameOptions: dict[str, str | bool] = response["PrivateDnsNameOptions"]
        self.MaintenanceOptions: dict[str, str] = response["MaintenanceOptions"]
        self.CurrentInstanceBootMode: str = response["CurrentInstanceBootMode"]
```
